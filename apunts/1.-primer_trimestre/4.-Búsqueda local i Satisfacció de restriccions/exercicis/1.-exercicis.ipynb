Haurás de definir una classe `ProblemaMotxilla` que hereti de `ProblemaBusquedaLocal` i implementi els mètodes necessaris per a la cerca local.

```python
import numpy as np


class ProblemaBusquedaLocal(object):
    """Aquesta és la classe abstracta per a un problema de cerca local.
    Per a implementar un problema de cerca local, subclassejar aquesta classe i
    implementar els mètodes estats_successors, funcio_avaluacio i es_solucio.
    """

    def __init__(self, inicial=None, **kwds):
        self.__dict__.update(inicial=inicial, **kwds)

    def estats_successors(self, estat):    raise NotImplementedError

    def es_solucio(self, estat):           raise NotImplementedError

    def funcio_avaluacio(self, state):     return NotImplementedError

    def __repr__(self):
        return '{}({!r})'.format(
            type(self).__name__, self.inicial)


# Definició de la classe ProblemaMotxilla que hereta de ProblemaBusquedaLocal
class ProblemaMotxilla(ProblemaBusquedaLocal):
    def __init__(self, items, capacity, inicial=None):
        super().__init__(inicial)
        self.items = items  # Llista de (weight, value) tuples
        self.capacity = capacity
        if inicial is None:
            # Estat inicial: cap element a la motxilla (tots False)
            self.inicial = [False] * len(self.items)

    def _get_current_weight_value(self, estat):
        current_weight = 0
        current_value = 0
        for i, included in enumerate(estat):
            if included:
                weight, value = self.items[i]
                current_weight += weight
                current_value += value
        return current_weight, current_value

    def estats_successors(self, estat):
        successors = []
        # Generar successors invertint l'estat d'un sol element
        for i in range(len(estat)):
            successor = list(estat) # Crear una còpia de l'estat actual
            successor[i] = not successor[i] # Invertir l'estat de l'element i
            successors.append(successor)
        return successors

    def es_solucio(self, estat):
        current_weight, _ = self._get_current_weight_value(estat)
        return current_weight <= self.capacity

    def funcio_avaluacio(self, estat):
        current_weight, current_value = self._get_current_weight_value(estat)
        if not self.es_solucio(estat):
            # Penalitzar fortament les solucions no factibles
            return 0  # O un número negatiu molt gran si busquem maximitzar
        return current_value

# Implementació de l'algoritme Hill Climbing
def hill_climbing(problema):
    current_state = problema.inicial
    current_value = problema.funcio_avaluacio(current_state)

    while True:
        neighbors = problema.estats_successors(current_state)
        best_neighbor = current_state
        best_neighbor_value = current_value

        for neighbor in neighbors:
            neighbor_value = problema.funcio_avaluacio(neighbor)
            if neighbor_value > best_neighbor_value:
                best_neighbor = neighbor
                best_neighbor_value = neighbor_value

        if best_neighbor_value <= current_value:
            # No hi ha un veí millor, hem arribat a un òptim local
            return current_state, current_value
        else:
            current_state = best_neighbor
            current_value = best_neighbor_value

# Exemple d'ús:
print("-----------------------------------------------")
print("Exemple d'ús del Hill Climbing per al problema de la motxilla:")

# Definir els elements (pes, valor)
items = [
    (2, 3), # item 0
    (3, 4), # item 1
    (4, 5), # item 2
    (5, 8), # item 3
    (9, 10) # item 4
]
capacity = 10

# Crear una instància del problema de la motxilla
problema_motxilla = ProblemaMotxilla(items, capacity)

# Executar Hill Climbing
final_state, final_value = hill_climbing(problema_motxilla)

print(f"Elements disponibles: {items}")
print(f"Capacitat de la motxilla: {capacity}")
print(f"Estat inicial (cap element): {problema_motxilla.inicial}")

selected_items = [items[i] for i, included in enumerate(final_state) if included]
selected_weights = sum(item[0] for item in selected_items)
selected_values = sum(item[1] for item in selected_items)

print(f"Estat final trobat per Hill Climbing: {final_state}")
print(f"Elements seleccionats: {selected_items}")
print(f"Pes total: {selected_weights}")
print(f"Valor total: {selected_values}")
print(f"Valor de la funció d'avaluació: {final_value}")

# Un altre exemple amb un òptim local diferent
print("\n-----------------------------------------------")
print("Un altre exemple (possiblement amb un òptim local diferent):")

items2 = [
    (1, 10), # item 0
    (3, 20), # item 1
    (2, 15), # item 2
    (5, 30)  # item 3
]
capacity2 = 6

problema_motxilla2 = ProblemaMotxilla(items2, capacity2)
final_state2, final_value2 = hill_climbing(problema_motxilla2)

selected_items2 = [items2[i] for i, included in enumerate(final_state2) if included]
selected_weights2 = sum(item[0] for item in selected_items2)
selected_values2 = sum(item[1] for item in selected_items2)

print(f"Elements disponibles: {items2}")
print(f"Capacitat de la motxilla: {capacity2}")
print(f"Estat final trobat per Hill Climbing: {final_state2}")
print(f"Elements seleccionats: {selected_items2}")
print(f"Pes total: {selected_weights2}")
print(f"Valor total: {selected_values2}")
print(f"Valor de la funció d'avaluació: {final_value2}")
```
